```python
basket= (
    df.groupby("Transaction ID")["Business_Group "]
      .apply(list)
      .reset_index()
)
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
te = TransactionEncoder()

te_array = te.fit(basket["Business_Group "]).transform(basket["Business_Group "])

basket_df = pd.DataFrame(
    te_array,
    columns=te.columns_
)
```

```python
basket_df.head()
```

```python
frequent_itemsets = apriori(
    basket_df,
    min_support=0.02,
    use_colnames=True
)
```

```python
frequent_itemsets
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="lift",
    min_threshold=1
)
```
